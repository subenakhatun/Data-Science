#### While loop
একটি if স্টেটমেন্টের কন্ডিশন সত্য হলে তার আওতাভুক্ত একটি কোড ব্লক রান করানো যায়। if এর ক্ষেত্রে কন্ডিশন সত্য হলে কোড ব্লকটি মাত্র একবার রান হয়। while লুপ মোটামুটি একইরকম ভাবে কাজ করে। যেমন - এরও একটি কন্ডিশন দরকার হয় এবং সেটি সত্য হলে, এর আওতাভুক্ত কোড রান করে। কিন্তু গুরুত্বপূর্ণ বিষয়টি হচ্ছে শুধু একবার না বরং অনেক বার রান করানো যায় (একে iteration বলে)।

অর্থাৎ, যতক্ষণ সেই while লুপের কন্ডিশন সত্য থাকে ততক্ষণ পর্যন্ত এর আওতাভুক্ত কোড রান করতেই থাকে। আর যখন কন্ডিশনটি মিথ্যা হয়ে যায় তখন while লুপের বাইরে গিয়ে প্রোগ্রামের পরবর্তী স্টেটমেন্ট গুলো রান করা শুরু করে।

প্রবলেম : বাস কাউন্টারের টিকিট বুকিং সিস্টেম
একটি বাসে মোট সিট আছে ১০টি। যতক্ষণ না বাসের সব সিট বুকড হয়ে যাচ্ছে (seats_available == 0), ততক্ষণ পর্যন্ত লুপটি চলবে এবং টিকিট বিক্রি করতে থাকবে।

শর্ত: * শুরুতে সিট সংখ্যা থাকবে ১০।

প্রতিবার লুপ চলার সময় ইউজারের কাছ থেকে ইনপুট নিতে হবে যে সে কয়টি টিকিট কিনতে চায়।

টিকিট কেনার পর মোট সিট থেকে তা বিয়োগ হবে।

যদি ইউজার এভেইলেবল সিটের চেয়ে বেশি টিকিট কিনতে চায় (যেমন সিট বাকি আছে ৩টি, কিন্তু ইউজার কিনতে চায় ৫টি), তাহলে স্ক্রিনে দেখাবে "দুঃখিত, এত সিট খালি নেই!" এবং টিকিট বুক হবে না।

সিট সংখ্যা ০ হয়ে গেলে লুপ বন্ধ হবে এবং দেখাবে "সব টিকিট বিক্রি হয়ে গেছে!"

```python
available_seat = 10
while available_seat > 0:
    buy_seat = int(input("Enter Numbe of seat you want to buy: "))
    if buy_seat > available_seat:
        print("Sorry have not sufficent seat!")
    else:
        available_seat = available_seat - buy_seat
        print("New Number of seat: ", available_seat)
    
print("All seats are sold out!")
```
